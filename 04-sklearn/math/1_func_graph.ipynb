```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 生成数据x、y
x = np.arange(0, 6.28, 0.01)
y = np.sin(x)
```

## 绘制函数图像

```python
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y=sin(x)")
plt.show()
```

## 绘制导函数图像

```python
y1 = np.cos(x)
plt.plot(x, y1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y'=cos(x)")
plt.show()
```

## 在同一张图中显示函数与导函数

```python
y2 = -np.sin(x)
plt.plot(x, y, label="f(x)")
plt.plot(x, y1, label="f'(x)", linestyle="--")
plt.plot(x, y2, label="f''(x)", linestyle="--")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y=sin(x)")
plt.legend()
plt.show()
```
