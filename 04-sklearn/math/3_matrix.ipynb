```python
import numpy as np
```

## 矩阵

```python
A = np.array([[1,2,3], [4,5,6]])
```

```python
print(A)
```

```python
print(A.shape)
```

```python
B = np.array([[1,3,7], [5,0,2]])
print(B)
```

```python
# 获取元素
print(A[0,2])
```

```python
print(A[0])
print(A[0][2])
```

```python
# 1. 转置
print(A.T)
```

```python
# 2. 矩阵相加
print(A + B)
```

```python
# 3. 矩阵与标量相乘
print(A * 3)
```

```python
# 4. 矩阵与向量相乘
# 4.1 对位相乘
x = np.array([2, 3, 4])
print(A * x)
```

```python
# 4.2 矩阵乘法
print(A @ x)
```

```python
# 5. 矩阵与矩阵相乘
# 5.1 对位相乘(阿达玛积)
print(A * B)
```

```python
# 5.2 矩阵乘法
print(A @ B.T)
```

```python
# 6. 逆
C = np.array([[1,2],[3,5]])
print(C)
```

```python
C_inv = np.linalg.inv(C)
C_inv = np.around(C_inv).astype(int)
print(C_inv)
```

```python
print(C @ C_inv)
print(C_inv @ C)
```

```python
# 7. 向量化
print(A.flatten())
print(B.flatten())
```

```python
# 8. 矩阵内积
print(np.sum(A * B))
```

```python
print(A.flatten() @ B.flatten())
```

```python
# 9. 克罗内克积
print(np.kron(A, B))
```
