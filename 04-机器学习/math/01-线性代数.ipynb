```python
import numpy as np
x  =  np.array([0,2,-1]) # 定义一个向量
# l0范数
l0_norm = np.sum(x != 0)
l0_norm = np.linalg.norm(x, ord=0)
```

```python
print("l1范数:", np.linalg.norm(x, ord=1))
```

```python
print("l2范数:", np.linalg.norm(x, ord=2))
```

```python
#矩阵的点击
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
inner_product = A@B
print("矩阵的内积:\n", inner_product)
```
