```python
import numpy as np
```

```python
x = np.array([2, -5, 8])
```

```python
print(x)
```

```python
print(x.shape)
```

```python
# 1. 转置
print(x.T)
```

```python
print(x.T.shape)
```

```python
# 2. 向量相加
y = np.array([-1, 3, 7])
print(x + y)
```

```python
# 3. 与标量相乘
print(x * 3)
```

```python
# 4. 与向量相乘(对位相乘)
print(x * y)
```

```python
# 5. 点积
print(x.dot(y))
```

```python
print(x @ y)
```

```python
# 6. 范数
norm = np.linalg.norm(x)
print(norm)
```

```python
norm0 = np.linalg.norm(x, ord=0)
print(norm0)
```

```python
norm1 = np.linalg.norm(x, ord=1)
print(norm1)
```

```python
norm2 = np.linalg.norm(x, ord=2)
print(norm2)
```

```python
norm_inf = np.linalg.norm(x, ord=np.inf)
print(norm_inf)
```
