```python
import numpy as np
```

```python
# 1. 均匀分布
# 1.1 小数
a = np.random.rand(3, 5)
print(a)
```

```python
# 1.2 整数
b = np.random.randint(0, 100, (3, 4))
print(b)
```

```python
# 2. 正态分布
# 2.1 标准正态分布
c = np.random.randn(5,2)
print(c)
```

```python
# 2.2 一般正态分布
d = np.random.normal(5, 2, (10,))
print(d)
```
