```python
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = np.array([10,20,30,40,50])
```

```python
# 均值
print(np.mean(x))
print(np.mean(y))
```

```python
# 方差
print(np.var(x))
print(np.var(y))
```

```python
# 标准差
print(np.std(x))
print(np.std(y))
```

```python
# 协方差
print(np.cov(x, y, bias=True))
```
