### 分组聚合计算

```python
df["month"] = pd.to_datetime(df["date"]).dt.to_period('M')
df
```

```python
# 按月分组
df_grouped = df.groupby("month")
print(df_grouped) #得到分组对象
```

```python
df_mean  = df_grouped[["temp_max", "temp_min"]].mean()
df_mean.head()
```

```python
# 分组频数计算 统计每个月的天气情况
df.groupby("month")["weather"].unique().head()
```

### 常用的排序方法

```python
# nlargest  按列排序的最大n个
# nsmallest  按列排序的最小n个
# sort_values 按列升序或者降序排列
# drop_duplicates 删除重复行
```

```python
df.nlargest(3, ['temp_max',"wind"]) # temp_max相同，wind小的在前
```

```python
df.nsmallest(3,['temp_max',"wind"])
```

```python
# 找出每年最高的温度
```

```python
df["year"] = pd.to_datetime(df["date"]).dt.to_period('Y')
df.nlargest(3, ['temp_max',"wind"])
```

```python
df.sort_values(by="temp_max", ascending=False).drop_duplicates(subset="year", keep="first")
```
