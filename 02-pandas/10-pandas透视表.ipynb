```python
# 什么是透视表？透视表是数据统计中的一个概念，可以快速统计出数据中某个字段的汇总值。
import pandas as pd
from pathlib import Path
source_path  = Path.cwd().parent / "资料-pd"

df = pd.read_csv(source_path / "sleep.csv")
df.head()
```

```python
sleep_duration_stage = pd.cut(df['sleep_duration'],[0, 5, 6, 7, 8, 9, 10, 11, 12])
```

```python
# 按睡眠时间划分
stress_level_stage = pd.cut(df['stress_level'],4)
```

```python
df.pivot_table(values='sleep_quality', index=[sleep_duration_stage, stress_level_stage], aggfunc='mean',columns=["occupation","gender"])
```

```python
df_date = pd.to_datetime(df['date'])
df["year"]  = df_date.dt.year
df["month"] = df_date.dt.month
df["day"] = df_date.dt.day
df["hour"] = df_date.dt.hour
df["minute"] = df_date.dt.minute
df
```

```python
# period  q1 q2 q3 ..
df['period'] = df['date'].dt.to_period('Q')
df
```

```python
# 当一个时间 - 另外一个时间 就是 timedelta64对象
timedelta = pd.Timestamp('2023-01-01') - pd.Timestamp('2022-01-01')
timedelta
```

```python
df["timedelta"] = pd.Timestamp('2016-01-01') - df["date"]
df
```
